```python
import os, subprocess, shutil, time
t0 = time.time()
os.makedirs('/kaggle/working', exist_ok=True)
cmd = """
set -x
nvidia-smi -L; echo "user=$(id -un) NVIDIA_DRIVER_CAPABILITIES=${NVIDIA_DRIVER_CAPABILITIES:-unset}"
command -v node || (curl -fsSL https://nodejs.org/dist/v22.11.0/node-v22.11.0-linux-x64.tar.xz | tar -xJ -C /opt && ln -sf /opt/node-v22.11.0-linux-x64/bin/* /usr/local/bin/)
node --version
curl -fsSL https://raw.githubusercontent.com/KealJones/squiggle/gpu-smoke/tests/eval/hf-gpu-smoke.sh -o /tmp/smoke.sh
bash /tmp/smoke.sh
"""
with open('/kaggle/working/smoke.log', 'w') as log:
    p = subprocess.Popen(['bash', '-c', cmd], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    for line in p.stdout:
        print(line, end='', flush=True); log.write(line); log.flush()
    code = p.wait()
for f in ('/tmp/probe.txt', '/tmp/case1.json'):
    if os.path.exists(f): shutil.copy(f, '/kaggle/working/')
print(f'exit {code} after {(time.time()-t0)/60:.1f} min')
```
